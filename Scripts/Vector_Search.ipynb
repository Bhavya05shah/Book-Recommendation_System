```python
from gradio_client.documentation import document
from langchain_classic import text_splitter
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
import pandas as pd
books_df = pd.read_csv("../Data/cleaned_data.csv")
```

```python
# books_df
```

```python
# .to_csv still applies the csv quotation rules, so any tabs, newlines and quotes can affect us in further
# processing, so we will avoid using .to_csv and also the delimiter sep "\n" contradicts with pythons
# books_df["description"].to_csv("tagged_descriptions.txt", sep="\t", index=False, header=False)

descriptions = (
    books_df["description"].astype(str)
    .str.replace(r"\s+", " ", regex=True)  # collapse newlines/tabs into spaces
    .str.strip()
)

with open("../Data/tagged_descriptions.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(descriptions))
```

```python
raw_docs = TextLoader("../Data/tagged_descriptions.txt").load()
# We use chunk_size as 0, so that each line is a single chunk and we dont want any data from the line 1 to
# transfer to line 2 so chunk overlap is 0, we could use chunk_size something if our descriptions were something
text_splitter = CharacterTextSplitter(chunk_size=1, chunk_overlap=0, separator="\n")
documents = text_splitter.split_documents(raw_docs)
```

#### all-MiniLM-L6-v2 is small and fast, and it runs fine on CPU. If you want better quality and don't mind a slightly slower run, try BAAI/bge-small-en-v1.5.

```python
embeddings = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
)
db_books = Chroma.from_documents(
    documents,
    embeddings = embeddings
)
```
